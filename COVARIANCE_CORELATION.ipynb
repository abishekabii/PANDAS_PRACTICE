```python
import pandas as pd
import seaborn as sns
```

```python
data = {
    'Tesla': [800, 820, 790, 830, 810],
    'Apple': [150, 155, 145, 160, 153]
}
```

```python
df=pd.DataFrame(data)
```

```python
df
```

```python
sns.pairplot(df)
```

```python
co_var=df.cov()
```

```python
co_var
```

```python
cov_value = co_var.loc['Tesla', 'Apple']
print("\nCovariance between Tesla & Apple:", cov_value)
```

It is +ve so both stocks are moving in same direction

```python
data2= {
    'Tesla': [800, 820, 790, 830, 810],
    'Apple': [150, 100, 185, 80, 110]
}
```

```python
df2=pd.DataFrame(data2)
df2
```

```python
co_var2=df2.cov()
co_var2
```

```python
cov_value2=co_var2.loc['Apple','Tesla']

print("\nCovariance between Tesla & Apple:", cov_value2)
```

It is -ve so both stocks are moving in opp direction

Now correlation

```python
co_rel1=df.corr()
co_rel1
```

```python
sns.pairplot(co_rel1)
```

```python
cor_val1=co_rel1.loc['Apple','Tesla']
print("\nCorelation between Tesla & Apple:", cor_val1)
```

```python
sns.pairplot(co_rel2)
```

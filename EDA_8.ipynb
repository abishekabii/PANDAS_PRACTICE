Univariate analysis 

Univariate analysis focuses on studying one variable to understand its characteristics.


Bivariate analysis 

Bivariate analysis focuses on exploring the relationship between two variables to find connections, correlations, and dependencies.

Multivariate analysis

Multivariate analysis examines the relationships between two or more variables in the dataset. It aims to understand how variables interact with one another.

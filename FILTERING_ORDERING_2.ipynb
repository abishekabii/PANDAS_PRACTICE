IS IN Function in pandas

The isin() method checks if the Dataframe contains the specified value(s).

It returns a DataFrame similar to the original DataFrame, but the original values have been replaced with True if the value was one of the specified values, otherwise False.


✔ Filtering based on multiple values
✔ Checking membership in lists/sets
✔ Works with both Series and DataFrames

```python
specific_countries=['India','Brazil']

df[df['Country'].isin (specific_countries)]
```

```python
df[df['Country'].isin('China')]
```

will throw an error or not work as expected because isin() expects a list, tuple, or set, but  provided a string ('China'),
So Give them in List / Set / Tuple

```python
df[df['Country'].isin(['China'])]
```

For a single value check we can use == operator also

```python
df[df['Country'] == 'China']
```

In pandas, the .str.contains() method is used to check if a string column contains a specific substring or pattern using regular expressions.

Syntax:
df['column'].str.contains('substring', case=True, na=False, regex=True)

Parameters:
substring → The text or pattern to search for.

case=True → Case-sensitive search (set False for case-insensitive).

na=False → Handles NaN values (treats them as False).

regex=True → Enables regular expressions.




```python
df[df['Capital'].str.contains('New')]
```

```python
df[df['Country'].str.contains("united")]
```

We are not mentioning case sensitive

```python
df[df['Country'].str.contains("united",case=False)]
```

```python
df[df['Country'].str.contains("United")]
```

Using Regex

```python
df[df['Country'].str.contains("^I",regex=True)]
```

By default it is true only

```python
df[df['Country'].str.contains("^Ir")]
```

To set index 

```python

df=pd.read_csv("FILES/world_population.csv")
df2=df.set_index('Country')
df2
```

Filtering based on Rows Using Filter function

```python
df2.filter(items=['Rank','Capital','Continent'])
```

By default it is having axis=1, we can also set axis

axis=0 → Works Vertically (Column-Wise) --> Affects Rows

axis=1 → Works Horizontally (Row-Wise) --> Affects Columns

```python
df2.filter(items=['Rank','Capital'],axis=1)
```

```python
df2.filter(items=['Zambia'],axis=1)
```

```python
df2.filter(items=['Zambia'],axis=0)
```

Using Like

```python
df2.filter(like='United',axis=1)
```

```python
df2.filter(like='United',axis=0)
```

loc --> Location --> Works on label index and integer also

```python
df2.loc['United States']
```

```python
df2.iloc['United States']
```

Sort values method

```python
df[df['Rank']<=10].sort_values(by='Rank')
```

By default it is ascending we can also sort descending

```python
df[df['Rank']<=10].sort_values(by='Rank',ascending=False)
```

Above we sorted single column , If we want to do for multiple column put it in list

```python
df[df['Rank']<=10].sort_values(by=['Rank','Country'])
```

But here we dont see that much effect so put capital first and then check

```python
df[df['Rank']<=10].sort_values(by=['Capital','Rank'])
```

We can also give ascending and descending to multiple columns

```python
df[df['Rank']<=10].sort_values(by=['Capital','Rank'],ascending=[False,True])
```

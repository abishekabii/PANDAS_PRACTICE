<i> One-Way ANOVA analysis</i> 

```python
import pandas as pd
from scipy import  stats
```

```python
data ={
    "AD_Type":["A","A","B","B","C","C"],
    "Sales":[100, 120, 95, 130, 110, 87]
}
```

```python
df = pd.DataFrame(data)
```

```python
df
```

```python
group_A = df[df['AD_Type'] == 'A']['Sales']
group_B = df[df['AD_Type'] == 'B']['Sales']
group_C = df[df['AD_Type'] == 'C']['Sales']
```

```python
group_A
```

```python
f_stat, p_value = stats.f_oneway(group_A, group_B, group_C)
print("F-statistic:", f_stat)
print("p-value:", p_value)
```

<i> Two-Way ANOVA analysis</i>

```python
import pandas as pd
import statsmodels.api as sm
from statsmodels.formula.api import ols

# Expanded dataset with replication (2 samples per group)
data = {
    'Ad_Type': ['A', 'A', 'A', 'A', 'B', 'B', 'B', 'B', 'C', 'C', 'C', 'C'],
    'Platform': ['TV', 'TV', 'Social Media', 'Social Media'] * 3,
    'Sales': [100, 102, 110, 108, 120, 118, 130, 128, 90, 92, 95, 93]
}
df = pd.DataFrame(data)

# Two-way ANOVA with interaction
model = ols('Sales ~ C(Ad_Type) + C(Platform) + C(Ad_Type):C(Platform)', data=df).fit()
anova_table = sm.stats.anova_lm(model, typ=2)
print(anova_table)
```

```python
from statsmodels.stats.weightstats import ztest as ztest
data=[88, 92, 94, 100, 83]
```

value is the mean

```python
ztest(data,value=100)
```

First is z_score,
 
Second is p_value

```python
data2=[88, 92, 94, 100, 83,99, 110, 102, 104, 114, 123]
```

```python
ztest(data2,value=110)
```

```python
ztest(data,data2)
```

Two sample T test

```python
from statsmodels.stats.weightstats import ttest_ind
```

```python
data3=[88, 92, 94, 100, 83,99, 110, 102, 104, 114, 123]
```

```python
ttest_ind(data3,data2)
```

First is t_statistic, 

Second is p_value, 

Thirs is degrees_of_freedom

```python
data4=[94, 104, 83,99, 110, 21, 14, 11, 123]
```

```python
ttest_ind(data4,data3)
```

```python
import pandas as pd
#import matplotlib.pyplot as plt
```

```python
df=pd.read_csv("FILES/Ice Cream Ratings.csv")
df
```

```python
df=df.set_index('Date')
df
```

Plot the graph using plot method

By default --> Line plot

```python
df.plot()
```

```python
df.plot(kind='line')
```

Subplot --> Dividing plot in to sub plots

```python
df.plot(subplots=True)
```

To Add title

```python
df.plot(title="ICE CREAM RATINGS")
```

Customize X and Y label

```python
df.plot(xlabel="DATE",ylabel="RATINGS SCORE")
```

Bar plot

```python
df.plot(kind='bar')
```

Stacked bar chart

```python
df.plot(kind='bar', stacked=True)
```

We can specify particular column also

```python
df['Flavor Rating'].plot(kind='bar')
```

By default this bar plot is vertical we can also make it as horizontal 

```python
df.plot(kind='barh')
```

```python
df.plot.barh()
```

Scatter plot

```python
df.plot(kind='scatter')
```

Showing error that scatter plot needs x and y columns

```python
df.plot(kind='scatter',x='Texture Rating',y='Overall Rating')
```

```python
df.plot(kind='scatter',x='Texture Rating',y='Overall Rating',s=300)
```

```python
df.plot(kind='scatter',x='Texture Rating',y='Overall Rating',s=300,c='red')
```

Histogram Plot

```python
df.plot(kind='hist')
```

```python
df.plot.hist()
```

We can specify the bins

```python
df.plot(kind='hist',bins=30)
```

Box Plot

```python
df.plot.box()
```

```python
df.plot(kind='box')
```

Area Plot

```python
df.plot(kind='area')
```

To change the size

```python
df.plot(kind='area',figsize=(10,5))
```

Pie chart

```python
df.plot(kind='pie')
```

Showing error needs to specify the column

```python
df.plot(kind='pie',y='Overall Rating')
```

Change the size also

```python
df.plot(kind='pie',y='Overall Rating',figsize=(10,7))
```
